## Part 1  -  Generating and plotting sinusoids

### Exercise 1  -  Time samples

Create a NumPy array with 100 equally spaced time samples over one second (0.00 to 0.99 s). Store it in a variable.

- Hint: `numpy.arange()` or `numpy.linspace(..., endpoint=False)`.

#### Solution

Create 100 equally spaced samples over one second. Both examples exclude the endpoint 0.00 to 0.99 s.

```python
import numpy as np
import matplotlib.pyplot as plt
```

#### Option A  -  `arange`

A 0.01 s step gives 100 samples.

```python
time = np.arange(0, 1, 0.01)
print(time)
```

#### Option B  -  `linspace`

`endpoint=False` gives the same samples.

```python
time = np.linspace(0, 1, 100, endpoint=False)  # Without endpoint=False, or endpoint=True, 1.00 would be included
print(time)
```

### Exercise 2  -  1 Hz sine

Create sinusoidal data with amplitude 1 and frequency 1 Hz at those time samples. Store it in `x1`. The time and signal arrays must have the same number of elements.

- Hint: a loop or list comprehension combined with `numpy.sin()`.

#### Solution

Use amplitude 1 and frequency 1 Hz $x(t)=\sin(2\pi t)$. The loop, list comprehension, and vectorized expression below are alternatives.

Store the first sinusoid in `x1`.

#### Angle check

Use `np.pi` for $\pi$ radians. The sine is zero mathematically. floating-point arithmetic gives a value very close to zero.

```python
print("π =", np.pi)
print("sin(π) =", np.sin(np.pi))  # Almost 0, floating-points can be funny like that
```

#### Option A  -  For-Loop

```python
x1 = np.array([])

for t in time:
    s = np.sin(t*2*np.pi)
    x1 = np.append(x1, s)

# Alternative way to do the same thing, using indexing instead of iterating over the array
# for i in range(len(time)):
#     s = np.sin(time[i] * 2 * np.pi)
#     x1 = np.append(x1, s)

plt.plot(time, x1)
plt.xlabel("Time (s)")
plt.ylabel("Amplitude")
plt.title("1 Hz sinusoid")
plt.show()
```

#### Option B  -  List comprehension

```python
x1 = [np.sin(t*np.pi*2) for t in time]  # This is a more Pythonic way to do the same thing as the for-loop above, using a list comprehension. It is also faster than the for-loop, but it takes some getting used to.
x1 = np.array(x1)

plt.plot(time, x1)
plt.xlabel("Time (s)")
plt.ylabel("Amplitude")
plt.title("1 Hz sinusoid")
plt.show()
```

#### Option C  -  NumPy expression

Apply `sin` to the whole array at once.

```python
x1 = np.sin(time*np.pi*2)
plt.plot(time, x1)
plt.xlabel("Time (s)")
plt.ylabel("Amplitude")
plt.title("1 Hz sinusoid")
plt.show()
```

## Part 2  -  Unit impulses and LTI systems

### Exercise 6  -  Unit impulse

Create a discrete-time unit impulse $\delta[n]$ with 10 samples.

- Create a sample-index array `n` from 0 to 9.
- Store the unit-impulse array in `unit_impulse`.
- All samples of the unit impulse are zero, except the sample at $n=0$, which is 1.
- Plot the unit impulse.

#### Solution

Create 10 sample indices. The unit impulse is zero everywhere except at $n=0$.

```python
n = np.arange(10)

unit_impulse = np.zeros(len(n))
unit_impulse[0] = 1

plt.stem(n, unit_impulse, basefmt=" ")
plt.xlabel("Sample index n")
plt.xticks(n)
plt.ylabel("Amplitude")
plt.title("Unit impulse")
```

### Exercise 7  -  Scale the impulse

Use the unit impulse from Exercise 6 to create

$$
x_1[n] = 2\delta[n].
$$

Plot $x_1[n]$.

- What changed compared with the unit impulse?

#### Solution

Multiplying the impulse by 2 changes its amplitude from 1 to 2.

```python
x1 = 2 * unit_impulse

plt.stem(n, x1, basefmt=" ")
plt.xlabel("Sample index n")
plt.xticks(n)
plt.ylabel("Amplitude")
plt.title(r"$x_1[n] = 2\delta[n]$")
```

### Exercise 8  -  Shift and scale the impulse

Create two shifted and scaled versions of the unit impulse

$$
x_2[n] = -\delta[n-2]
$$

and

$$
x_3[n] = 3\delta[n-5].
$$

Plot both signals.

- At which sample does each impulse occur?
- What determines the position of an impulse?
- What determines its amplitude?

Use the existing `unit_impulse` array shift it with zero padding, then scale it.

#### Solution

A shift changes the sample at which the impulse occurs. Scaling changes its amplitude.

Prepend zeros and remove the same number of trailing samples to shift `unit_impulse` without wrapping. Then apply the gain.

```python
x2 = -np.concatenate((np.zeros(2), unit_impulse[:-2]))
x3 = 3 * np.concatenate((np.zeros(5), unit_impulse[:-5]))

plt.stem(n, x2, linefmt="C0-", markerfmt="C0o", basefmt=" ")
plt.stem(n, x3, linefmt="C1-", markerfmt="C1o", basefmt=" ")
plt.xlabel("Sample index n")
plt.xticks(n)
plt.ylabel("Amplitude")
plt.title("Shifted and scaled impulses")
plt.legend([r"$x_2[n]$", r"$x_3[n]$"])
```

### Exercise 10  -  From impulse response to output

An LTI system has the following impulse response

$$
h[n] = \delta[n] + 0.5\delta[n-1].
$$

Store $h[n]$ in `impulse_response` and plot it.

Use the signal from Exercise 9 as the input

$$
x[n] = 2\delta[n] - \delta[n-2] + 3\delta[n-5].
$$

For each impulse in $x[n]$, create the corresponding scaled and shifted copy of $h[n]$. Add these copies to obtain the output $y[n]$.

Plot $y[n]$.

- Why can the output of an LTI system be constructed in this way?
- Do not use `np.convolve()` in this exercise.

#### Solution

For an LTI system, scaling and shifting an impulse causes the impulse response to be scaled and shifted in the same way. Additivity then allows the output components to be added.

```python
impulse_response = np.zeros_like(unit_impulse)
impulse_response[0] = 1
impulse_response[1] = 0.5

plt.stem(n, impulse_response, basefmt=" ")
plt.xlabel("Sample index n")
plt.xticks(n)
plt.ylabel("Amplitude")
plt.title("Impulse response")
```

The three input components produce $2h[n]$, $-h[n-2]$, and $3h[n-5]$.

```python
y1 = 2 * impulse_response

y2 = np.zeros_like(unit_impulse)
y2[2:] = -impulse_response[:-2]

y3 = np.zeros_like(unit_impulse)
y3[5:] = 3 * impulse_response[:-5]

y = y1 + y2 + y3

plt.stem(n, y, basefmt=" ")
plt.xlabel("Sample index n")
plt.xticks(n)
plt.ylabel("Amplitude")
plt.title("Output signal")

print(y)
```

The output can be constructed this way because the system is linear and time invariant scaling, shifting, and adding input impulses produces the same scaling, shifting, and adding of their impulse responses.
